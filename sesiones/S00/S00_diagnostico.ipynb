## Bloque 2 — pandas mínimo (10 min)

```python
import pandas as pd

# Consigue los datos del repositorio del curso (una vez por sesión de Colab).
ARCHIVO = "incidentes_medellin_muestra_5000.csv"
URL_REPO = "https://raw.githubusercontent.com/Nitobest/curso-aad-python/main/datos/" + ARCHIVO

def asegurar_datos(archivo, url_repo):
    from pathlib import Path
    import requests
    if Path(archivo).exists():
        return True
    try:
        r = requests.get(url_repo, timeout=120)
        if r.ok:
            Path(archivo).write_bytes(r.content)
            return True
    except Exception:
        pass
    print("⚠️ No pude bajar el archivo del repositorio. Use el Plan B (más abajo).")
    return False

asegurar_datos(ARCHIVO, URL_REPO)
URL = ARCHIVO
```

Los nombres de columna vienen como en el archivo original: mayúsculas, espacios, tildes. La siguiente celda los pasa a minúsculas con guion bajo (en la Sesión 1 vemos cómo funciona); por ahora solo ejecútela.

```python
import unicodedata

def limpiar_nombre(nombre):
    nombre = unicodedata.normalize("NFKD", nombre.strip().lower()).encode("ascii", "ignore").decode()
    return nombre.replace(" ", "_")

df.columns = [limpiar_nombre(c) for c in df.columns]
df = df.rename(columns={"ano": "anio", "fecha_incidente": "fecha", "hora_incidente": "hora", "clase_incidente": "clase",
                        "gravedad_incidente": "gravedad", "comuna_planeacion": "comuna", "barrio_planeacion": "barrio"})
list(df.columns)
```
